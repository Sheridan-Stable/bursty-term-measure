```python
import json
from keybert import KeyBERT
```

```python
infile_path = '../1-preprocessing/GENIAcorpus3.02-preprocessed-merged.json'

with open(infile_path, 'r') as file:
    # Load the JSON contents into a Python variable
    data = json.load(file)

```

```python
kw_model = KeyBERT()
for doc in data:
   print("------------------------------------------------------------------------")
   bag_of_words = kw_model.extract_keywords(doc, keyphrase_ngram_range=(1, 1), stop_words='english',top_n=25, highlight=True,  use_maxsum=True, nr_candidates=25)
   for bag in bag_of_words:
         print(bag[0],bag[1] )
```
