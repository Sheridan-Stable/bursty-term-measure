```python
# Imports
# !pip install keybert PS@R: Don't do this. Do it through requirements.txt
import json
from keybert import KeyBERT
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Path to GENIA corpus abstracts
infile_path = '../1-preprocessing/GENIAcorpus3.02-preprocessed-merged.json'

# Load the JSON contents into a Python variable
with open(infile_path, 'r') as file:
    docs = json.load(file)

# Check abstracts
print(docs)
```

```python
# Extract keywords with KeyBERT
# This vectorizer ensure that KeyBERT recognizes hyphens
#vectorizer = CountVectorizer(token_pattern=r'(?u)\b[\w-]+\b')
kw_model = KeyBERT()
for doc in docs:
   print("------------------------------------------------------------------------")
   #doc_processed = text.replace("-", "__")
   #keywords = kw_model.extract_keywords(doc, keyphrase_ngram_range=(1, 1), stop_words='english',top_n=25, highlight=True, vectorizer=vectorizer)
   keywords = kw_model.extract_keywords(doc, keyphrase_ngram_range=(1, 1), stop_words='english',top_n=25, highlight=True)
   for keyword in keywords:
        print(keyword[0], keyword[1])
```
