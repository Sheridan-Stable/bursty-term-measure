```python
# Imports
# !pip install keybert PS@R: Don't do this. Do it through requirements.txt
import json
from keybert import KeyBERT
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Path to GENIA corpus abstracts
infile_path = '../1-preprocessing/GENIAcorpus3.02-preprocessed-merged.json'

# Load the JSON contents into a Python variable
with open(infile_path, 'r') as file:
    docs = json.load(file)

# Check abstracts
print(docs)

number_of_docs = len(docs)
```

```python
# Extract keywords with KeyBERT
# This vectorizer ensure that KeyBERT recognizes hyphens
vectorizer = CountVectorizer(token_pattern=r'(?u)\b[\w-]+\b')
kw_model = KeyBERT()
list_of_keyword = []
for doc in docs:
   print("------------------------------------------------------------------------")
   #doc_processed = text.replace("-", "__")
   keywords_and_score = kw_model.extract_keywords(doc, keyphrase_ngram_range=(1, 1), stop_words='english',top_n=25, highlight=True, vectorizer=vectorizer)
   # keywords = kw_model.extract_keywords(doc, keyphrase_ngram_range=(1, 1), stop_words='english',top_n=25, highlight=True)

   keywords =[]
   for keywords_and_score in keywords_and_score:
        keywords.append(keywords_and_score[0])
   list_of_keyword.append(keywords) 

print(len(list_of_keyword))
```

```python
def count_docs_with_keyword(keyword,list_of_doc):
    # Count number of documents where t is selected by KeyLLM
    return  sum(1 for keywords in list_of_doc if keyword in keywords)
    
def ranking(number_of_documents, count):
    return (1/number_of_documents)*count

```

```python
for doc in list_of_keyword:
    for keywords in doc:
        count = count_docs_with_keyword(keywords,doc)
        rank= ranking(number_of_docs,count)
        print(keywords,rank )
```
